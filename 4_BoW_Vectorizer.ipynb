```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Step 1: Sample text data
corpus = [
    "NLP is fun",
    "NLP is powerful",
    "AI is fun"
]
```

```python
# Step 2: Initialize BoW model
vectorizer = CountVectorizer()
```

```python
# Step 3: Fit and transform text
X = vectorizer.fit_transform(corpus)
```

```python
# Step 4: Show vocabulary
print("Vocabulary:", vectorizer.get_feature_names_out())
```

```python
# Step 5: Convert to array (matrix)
print("\nBoW Matrix:\n", X.toarray())
```
