```python
import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab')

from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer, RegexpStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
text = "The researchers were analyzing the studies related to connected systems"
tokens = word_tokenize(text.lower())
tokens
```

```python
# Porter stemmer

porter = PorterStemmer()
[porter.stem(word) for word in tokens]
```

```python
# Lancaster Stemmer
lancaster = LancasterStemmer()
[lancaster.stem(word) for word in tokens]
```

```python
# Snowball stemmer
snowball = SnowballStemmer("english")
[snowball.stem(word) for word in tokens]
```

```python
# regex stemmmer
regexp = RegexpStemmer('ing$|ed$|s$')
[regexp.stem(word) for word in tokens]
```

```python
# wordnet lemmatizer
lemmatizer = WordNetLemmatizer()
[lemmatizer.lemmatize(word,pos='v') for word in tokens]
```
