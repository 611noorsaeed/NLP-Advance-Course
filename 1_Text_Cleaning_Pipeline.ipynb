# Step 1: Lower Casing

```python
text = "I Love To Play with my Two Dogs and Two Cats"
text
```

# Step 2: Remove Everything Except (Characters / Digits)

```python
text = "i love to *)*() play with my 2 cats and 2 #@$#@$$ dogs"
text
```

```python
import re
text = re.sub(r"[^a-zA-Z0-9\s]","",text)
text
```

# Step 3: Tokenization

```python
text = text.split()
text
```

# Step 4: Stopwords removal

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = stopwords.words("english")
print(stop_words[:5])
print(len(stop_words))
```

```python
text = [w for w in text if w not in stop_words]
text
```

# Step 5: Stemming Lemmatization

```python
text = "I am having food since morning"
text
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
text = [stemmer.stem(w) for w in text.split()]
text
```
